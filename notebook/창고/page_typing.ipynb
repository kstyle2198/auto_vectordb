```python
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b" , temperature=0)
llm
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
system_prompt = """
당신은 마크다운 문서 페이지 구조 분석 전문가입니다.
입력된 마크다운 텍스트를 분석하여 전체 페이지의 성격을 다음 세 가지 중 하나로 분류하세요:

1. 제목 (Title)
2. 목차 (Table of Contents)
3. 내용 (Content)

[분류 정의]

- 제목 (Title):
  - 문서 전체 또는 페이지 단위를 대표하는 헤드라인
  - 여러 개의 헤더가 있더라도, 
    ▶ 문서 주제 선언
    ▶ 연도 / 버전 / 규칙명 / 가이드명
    ▶ 서문(Introduction) 성격
    을 나타내면 '제목'으로 분류
  - 하위 설명이나 본문이 없고, 구조 안내 목적이 명확하지 않으면 제목으로 간주

- 목차 (Table of Contents):
  - 문서 내부 이동을 목적으로 한 구조적 안내 페이지
  - 다음 특징이 **명확히 존재할 때만** 목차로 분류:
    ▶ 섹션 번호 체계 (1, 1.1, 2.3 등)
    ▶ 링크 표현 (#, [], page, 이동, 참조 등)
    ▶ "목차", "Contents", "Table of Contents" 명시
    ▶ 다수의 항목이 동일한 형식으로 나열됨
  - 단순 헤더 나열이나 연도/소개 문구는 목차가 아님

- 내용 (Content):
  - 문장형 설명, 규칙, 표, 데이터, 절차, 조건 등 구체적 정보가 포함된 경우
  - 한 줄이라도 설명 문장이 있으면 내용으로 분류

[중요 판단 규칙]

- 짧은 헤더 여러 개 ≠ 목차
- 링크/이동 목적이 명확하지 않으면 목차로 분류하지 말 것
- 제목과 목차가 헷갈릴 경우 **항상 제목을 우선 선택**

출력은 부연설명 없이 반드시 다음 중 하나만 반환하세요:
'제목', '목차', '내용'
"""

agent = create_agent(
    llm,
    system_prompt=system_prompt,
)

doc = """

## 2025

## Introduction to the Classification Technical Rules
"""

query = f"""아래 페이지를 분류해주세요. (제목, 목차, 내용)
{doc}

"""
inputs = {"messages": [HumanMessage(content= query)]}
res = agent.invoke(inputs)
Markdown(res["messages"][-1].content)
```

```python
def page_category(page_content):

    ## 첫줄 삭제
    page_content = page_content.replace("This page explains Introduction to the Classification Technical Rules_2025 that belongs to Rule,KR categories.", "")
    
    query = f"""아래 페이지를 분류해주세요. (제목, 목차, 내용)
    {page_content}
    """
    inputs = {"messages": [HumanMessage(content= query)]}
    res = agent.invoke(inputs)
    return res["messages"][-1].content
```

```python
results = []
for idx, doc in enumerate(loaded_data):
    
    res = page_category(doc.page_content)
    print(f"====={res}=====")
    print(doc.page_content)
    results.append(res)

    if idx == 10:
        break
```
